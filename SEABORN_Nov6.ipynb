```python
dataset = sns.load_dataset('iris')
dataset
sns.lineplot(x="sepal_length",y="sepal_width",data=dataset)
```

```python
sns.barplot(x='species',y='sepal_length',data=dataset)
```

```python
sns.barplot(x='species',y='sepal_length',palette= 'coolwarm',data=dataset)
```

```python
#Count Plot
data = sns.load_dataset('tips')
sns.countplot(x='sex',data = data)
```

```python
sns.boxplot(x='sex',y ='total_bill',data=data)
```

```python
sns.boxplot(x='sex',y ='total_bill',palette='rainbow',data=data)
```

```python
#orient
sns.boxplot(data = data,palette='rainbow',orient='h')
```

```python
#hue
sns.boxplot(x='day',y='total_bill',hue='sex',data=data,palette='coolwarm')
```
