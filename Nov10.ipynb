```python
import seaborn as sns
import matplotlib.pyplot as plt
# Loading dataset
data = sns.load_dataset("iris")

# creating a violin plot
sns.violinplot(x='species', y='sepal_width',data=data,hue= 'petal_length',palette = 'coolwarm')
data
```

```python
data = sns.load_dataset("tips")
sns.violinplot(x='day',y='total_bill',data = data,hue='sex',palette='Set1')
```

```python
#STRIP PLOT
data = sns.load_dataset('iris')
sns.stripplot(x='species',y='sepal_width',data=data,palette='Set1')
```
